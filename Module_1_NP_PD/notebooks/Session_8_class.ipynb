### Concept notes: Lesson 2
**Problem:** `"₹1,099"` is text. Text can't be averaged, sorted by value or compared. We must turn it into a number.

**Step by step on one value** (a different example, so you apply it yourself):
```python
s = "$2,500"
s = s.replace("$", "")      # "2,500"   replace(old, new); new = "" means delete
s = s.replace(",", "")      # "2500"
n = float(s)                # 2500.0   float() converts text to a decimal number
```
**On a whole column** you can't call `replace` directly. Put `.str` first:
```python
df["salary"].str.replace("$", "")        # applies to every value in the column
df["salary"].str.replace("$", "").str.replace(",", "")   # chain: output feeds the next call
df["salary"].str.replace("$", "").str.replace(",", "").astype(float)   # whole column to numbers
```
- `.str` = "treat each value as text". `astype(float)` = "convert the column's type".
- Missing values (`NaN`) stay `NaN`, they don't break `astype(float)`.

**Keeping the result:** these lines create a new Series and leave the original untouched. To keep it, assign back: `df["salary"] = ...`.

**Checking it worked:** `df["salary"].dtypes` should say `float64`, and `describe()` now includes the column.

**Run-once warning:** after the column is numbers, `.str` no longer works on it. Task 5 shows what that looks like.

### Concept notes: data audit
Before cleaning, find *which* columns are dirty. One question drives everything: **does the data type match what the values mean?** A price, rating or count should be numeric. If it is text, it is a suspect.

| Command | What it reveals |
|---|---|
| `df.dtypes` | Type of every column (`float64`, `int64`, `str`). |
| `df.select_dtypes(exclude="number")` | Only the non-numeric columns, so you can review them quickly. |
| `df.sample(5)` | 5 random rows. Better than `head()` on big tables (the top can look cleaner than the rest). |
| `df["c"].str.contains("x")` | True/False per row for rows containing `x`. `.sum()` counts them. Add `na=False` so missing values count as False instead of causing trouble. |
| `pd.to_numeric(df["c"], errors="coerce")` | Tries to make numbers, and turns anything impossible into `NaN` instead of crashing. |

**The `to_numeric` trick:** compare the NaN count before and after. If it jumps from 8 to 1465, nearly every value failed to convert, so the column is full of non-numeric characters. If it does not move, the column is already clean.

Also useful on text columns: `df["c"].unique()` and `df["c"].value_counts()` reveal typos and stray spaces (`"Electronics"` vs `"electronics "`).

### Concept notes: `str.contains` and `to_numeric` (extra drill)
Both tools answer one question: *what is making this column dirty?*

**`.str.contains("x")`** asks each row "do you contain x?" and returns True/False. Chain `.sum()` to count, or use it as a filter.
```python
names = pd.Series(["Red Pen (Pack of 2)", "Blue Pen", None, "Black Marker (Pack of 5)"])
names.str.contains("Pack", na=False)          # [True, False, False, True]
names.str.contains("Pack", na=False).sum()    # 2
names[names.str.contains("Pack", na=False)]   # only the matching rows
```
**`pd.to_numeric(s, errors="coerce")`** converts to numbers and turns anything impossible into NaN instead of crashing.
```python
s = pd.Series(["10", "20", "1,500", "N/A"])
pd.to_numeric(s, errors="coerce")             # [10, 20, NaN, NaN]
pd.to_numeric(s, errors="coerce").isna().sum()  # 2
```
Trick: compare the NaN count **before and after**. A big jump means many values were not clean numbers. No jump means the column is already clean. Rows that became NaN are exactly the dirty ones, and you can show them with `s[converted.isna()]`.

## Lesson 3: Merge (joining two tables)

### Concept notes: what a merge does and the four join types
A merge glues two tables side by side by matching rows on a **key column** (here `dept_id`). The `how=` argument decides what happens to rows that have **no match** on the other side.

```python
emp  = pd.DataFrame({"emp": ["Asha", "Ben", "Chen"], "dept_id": [1, 2, 3]})
dept = pd.DataFrame({"dept_id": [1, 2, 4], "dept": ["HR", "IT", "Ops"]})
emp.merge(dept, on="dept_id", how="inner")
```
Chen (dept 3) has no department row. Ops (dept 4) has no employee. Same two tables, four different answers:

| `how=` | Keeps | Rows | Where NaN appears |
|---|---|---|---|
| `inner` | only keys found in **both** | 2 (Asha, Ben) | nowhere |
| `left` | **every row of the left** table, plus matches | 3 (Asha, Ben, Chen) | `dept` is NaN for Chen |
| `right` | **every row of the right** table, plus matches | 3 (Asha, Ben, Ops) | `emp` is NaN for Ops |
| `outer` | **everything** from both tables | 4 | both sides, wherever there was no match |

**Rule of thumb:** left = "keep all my main rows and attach extra info" (all orders, attach product details). Inner = "only rows I can fully explain". Right is just left with the tables swapped. Outer = "show me everything, including the mismatches".

**Predict before you run.** Count the keys on each side, decide which rows survive, then check with `len()`. This habit catches most merge bugs.
