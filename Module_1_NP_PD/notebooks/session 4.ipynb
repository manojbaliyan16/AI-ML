### Visualizing One-Hot Encoding

Let's trace exactly how our input array `[1, 3, 4, 1, 2]` translates into coordinates inside our $5 \times 5$ matrix:

* **Input Array:** `arr = [1, 3, 4, 1, 2]` (length $N=5$, max value is $4$ so columns $C=5$)
* **Initial Template Matrix:** A $5 \times 5$ grid of all zeros.

Here is how we map each element to a coordinate `(row_index, column_index)`:

```text
Element Index (Row)   Value (Col Index)     Coordinate -> Set to 1
       0           ->         1         ->     (0, 1)
       1           ->         3         ->     (1, 3)
       2           ->         4         ->     (2, 4)
       3           ->         1         ->     (3, 1)
       4           ->         2         ->     (4, 2)
```

### Step-by-Step Matrix Mapping:

```text
          Col 0  Col 1  Col 2  Col 3  Col 4
Row 0 -> [  0,     1,     0,     0,     0  ]   (Value 1 at Row 0)
Row 1 -> [  0,     0,     0,     1,     0  ]   (Value 3 at Row 1)
Row 2 -> [  0,     0,     0,     0,     1  ]   (Value 4 at Row 2)
Row 3 -> [  0,     1,     0,     0,     0  ]   (Value 1 at Row 3)
Row 4 -> [  0,     0,     1,     0,     0  ]   (Value 2 at Row 4)
```

````python
import numpy as np

def convert_to_one_hot(arr):
    # Convert input list/array to a 1D NumPy integer array
    arr = np.asarray(arr, dtype=int)

    # Determine dimensions
    N = len(arr)
    C = np.max(arr) + 1

    # Initialize an all-zeros array
    one_hot = np.zeros((N, C), dtype=int)

    # Use fancy indexing to set the appropriate elements to 1
    one_hot[np.arange(N), arr] = 1

    return one_hot

# Test with the sample input
sample_input = [1, 3, 4, 1, 2]
output_matrix = convert_to_one_hot(sample_input)
print(output_matrix)
````

### Understanding `np.zeros()`

`np.zeros()` is a NumPy function that creates a new array of a specified shape, where every single element is initialized to `0`.

#### 1. The Shape Parameter
To define the dimensions of your array, you pass a **tuple** representing `(rows, columns)` (or just a single integer for a 1D array).
* `np.zeros(5)` creates a 1D array with 5 zeros: `[0., 0., 0., 0., 0.]`
* `np.zeros((3, 4))` creates a 2D matrix with 3 rows and 4 columns.

#### 2. The Data Type (`dtype`)
By default, NumPy creates arrays with floating-point numbers (`float64`), which is why you might see decimals like `0.`.

If you want integers (like we did for our one-hot encoded matrix), you specify `dtype=int`:
* `np.zeros((2, 3), dtype=int)`

```text
Initial Template Matrix:
[[0, 0, 0],
 [0, 0, 0]]
```

### Demystifying `one_hot[np.arange(N), arr] = 1`

When you pass two lists (or arrays) inside the square brackets `matrix[rows, cols]`, NumPy does not look at them as a block. Instead, it pairs them up **one-by-one** based on their positions (index-wise) to form coordinate pairs `(row, col)`.

Let's break down our two arrays of length $N=5$:
1. `np.arange(5)` generates: `[0, 1, 2, 3, 4]` (Your row coordinates)
2. `arr` contains: `[1, 3, 4, 1, 2]` (Your column coordinates)

#### The Zip-like Pairing Process:
NumPy aligns these two arrays side-by-side and pairs them up at each index:

```text
Index:      0      1      2      3      4
Rows:      [0,     1,     2,     3,     4]
            |      |      |      |      |
Cols:      [1,     3,     4,     1,     2]
            v      v      v      v      v
Pairs:    (0,1)  (1,3)  (2,4)  (3,1)  (4,2)
```

#### The Assignment:
When we set `one_hot[np.arange(N), arr] = 1`, NumPy goes to those exact coordinates and updates their values to `1` in a single, lightning-fast step:

* `one_hot[0, 1] = 1`
* `one_hot[1, 3] = 1`
* `one_hot[2, 4] = 1`
* `one_hot[3, 1] = 1`
* `one_hot[4, 2] = 1`

All other coordinates remain untouched as the original `0`s we initialized with `np.zeros()`!

### 📓 Study Note: Mastering NumPy `np.pad()`

Padding is used to add a surrounding border of values around an array. It modifies the shape of an $N$-dimensional array by adding elements to the boundaries.

---

### 1. How `pad_width` Controls the Border

`pad_width` determines how many elements are added to each edge of each dimension.

#### A. Symmetric / Uniform Padding (1D & 2D)
* **`pad_width=1`**:
  Adds exactly **1 element** to every boundary of every axis (top, bottom, left, right).
  * Example: A $(3 	imes 3)$ matrix becomes $(5 	imes 5)$.

#### B. Axis-Specific Padding (2D Matrix)
To pad dimensions differently, pass a tuple of pairs: `((top, bottom), (left, right))`.

* **`pad_width=((2, 1), (0, 3))`**:
  * **Vertical Axis (Rows)**: Adds **2 rows** to the top, **1 row** to the bottom.
  * **Horizontal Axis (Columns)**: Adds **0 columns** to the left, **3 columns** to the right.

---

### 2. Common Padding Modes (`mode`)

The `mode` parameter determines *what values* occupy the padded margins:

| Mode | Behavior | Example (`[1, 2, 3]` padded by 1) |
| :--- | :--- | :--- |
| `'constant'` | Fills with a static value (use `constant_values=val`, default is `0`). | `[0, 1, 2, 3, 0]` |
| `'edge'` | Extends the outermost values outward. | `[1, 1, 2, 3, 3]` |
| `'reflect'` | Mirrors values across the edge (excluding the edge itself). | `[2, 1, 2, 3, 2]` |
| `'symmetric'` | Mirrors values across the edge (including the edge itself). | `[1, 1, 2, 3, 3]` |
| `'wrap'` | Wraps values around to the opposite side. | `[3, 1, 2, 3, 1]` |

````python
def rearrange_image_sections(data, num_splits):
    # Step 1: Split the 'data' array horizontally into 'num_splits' sections
    # Hint: Remember which axis controls horizontal (row-wise) splits
    sections = np.split(data,num_splits,axis=0)

    # Step 2: Rearrange the list of sections so the last one is at the beginning
    # Hint: Use Python list slicing! (e.g., sections[-1:] and sections[:-1])
    rearranged_sections = sections[-1:] + sections[:-1]

    # Step 3: Concatenate the rearranged list of sections back into a single array
    # Hint: Think about which axis we are joining them along
    output_array = np.concatenate(rearranged_sections, axis=0)

    return output_array

# Test your function here with a simple mock 2D array!
mock_data = np.arange(12).reshape(6, 2)
print("Original Array:\n", mock_data)
# try rearranging with 3 splits
print("\nRearranged Array:\n", rearrange_image_sections(mock_data, 3))
````
