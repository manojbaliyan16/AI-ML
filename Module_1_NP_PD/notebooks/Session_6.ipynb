## Demo A: the `how=` parameter

`how` decides which rows survive when an ID is in only one table:

| how | keeps |
|---|---|
| `'inner'` (default) | only IDs found in **both** tables |
| `'left'` | **every row of the left table**, plus matches from the right |
| `'right'` | **every row of the right table**, plus matches from the left |
| `'outer'` | **every row from both tables** |

Where there is no match, the missing side is filled with `NaN`.

```python
pd.merge(products, reviews, on='product_id', how='inner')
```

```python
pd.merge(products, reviews, on='product_id', how='left')
```

```python
pd.merge(products, reviews, on='product_id', how='right')
```

```python
pd.merge(products, reviews, on='product_id', how='outer')
```

```python
stock = pd.DataFrame({
    'item_id': [101, 103, 105, 107],
    'units':   [10, 0, 25, 8],
})

pd.merge(products, stock, left_on='product_id', right_on='item_id', how='left')
```
